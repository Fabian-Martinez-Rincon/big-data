### Enunciado
Ha sido contratado para el análisis de uso de un videojuego online donde millones de jugadores se conectan a diario para combatir con otros jugadores. En un combate, un jugador “retador” elije un oponente (jugador “retado”) y como resultado del combate el retador obtiene un puntaje y el tiempo en segundos que duró el combate. Por cada combate, los servidores del videojuego almacena la siguiente información:

- ID_Jugador_Retador
- ID_Jugador_Retado
- Puntaje obtenido (por el retador)
- Tiempo del combate en segundos

Todos los jugadores pueden participar cuántas veces deseen y por cada combate obtienen un puntaje y también se registra el tiempo que duró el combate. Un jugador puede combatir todas las veces que quiera.

Un fragmento del dataset de este jugador podría ser:

| 231            | 98                 | 789    | 302    |

Semanalmente los administradores del juego quieren obtener algunas estadísticas para regalarle premios a los mejores jugadores y armar el ranking semanal. Al finalizar la semana se desea saber:
### 1) El jugador más “retador” y el jugador más “retado”.

---

**Explicación**

En este ejemplo se utiliza el framework `MRE` para ejecutar un **Job MapReduce** que procesa los combates de jugadores.

- En la **fase Map**, se identifican los jugadores que aparecen como *retadores* y *retados* en cada línea de entrada.  
- En la **fase Reduce**, se cuentan cuántas veces aparece cada jugador y se determina cuál fue el más frecuente dentro de cada grupo.  

De esta forma, el job permite identificar el jugador con mayor participación en los combates, ya sea como retador o como retado.

### Dag



```python
import sys
sys.path.append("..") 
from MRE import Job

inputDir = "../Datasets/TP1/input/"
outputDir = "../Datasets/TP1/out_punto1/"

def fmap(key, value, context):
    cols = value.strip().split()
    id_retador = key
    id_retado  = cols[0]
    context.write("RETADOR", id_retador)
    context.write("RETADO",  id_retado)

def fred(key, values, context):
    conteos = dict()
    for jugador in values:
        conteos[jugador] = conteos.get(jugador, 0) + 1
    jugador_max = max(conteos, key=conteos.get)
    context.write(key, (jugador_max, conteos[jugador_max]))

job = Job(inputDir, outputDir, fmap, fred)
success = job.waitForCompletion()
```

### 2) El jugador que más puntos obtuvo en promedio

$$
PP_i = \frac{puntaje\_total\_de\_todos\_los\_combates_i + 1}{cantidad\_de\_combates\_como\_retador_i + 1}
$$

*(el +1 en el numerador y en el denominador es para evitar divisiones por cero en el caso que un jugador no haya “retado” a nadie, en cuyo caso tendrá el puntaje mínimo de 1).*

---

**Explicación**

En este job se busca calcular el **puntaje promedio** obtenido por cada jugador que actúa como *retador*.

- En la **fase Map**, se emiten pares con el ID del retador y su puntaje junto a un contador de combates.  
- En la **fase Reduce**, se acumulan los puntajes y combates para cada jugador y se aplica la fórmula:

$$
\text{Promedio} = \frac{\text{total\_puntos} + 1}{\text{total\_combates} + 1}
$$

El término `+1` se utiliza para evitar divisiones por cero y suavizar el cálculo cuando hay pocos combates.

### Dag




```python
import sys
sys.path.append("..") 
from MRE import Job

inputDir = "../Datasets/TP1/input/"
outputDir = "../Datasets/TP1/out_punto2/"

def fmap2(key, value, context):
    cols = value.strip().split()
    id_retador = key
    puntaje = int(cols[1]) if len(cols) == 3 else int(cols[2])
    context.write(id_retador, (puntaje, 1))

def fred2(key, values, context):
    total_puntos, total_combates = 0, 0
    for puntos, combates in values:
        total_puntos += puntos
        total_combates += combates
    # fórmula con +1
    promedio = (total_puntos + 1) / (total_combates + 1)
    context.write("PROMEDIO", (key, promedio))

job = Job(inputDir, outputDir, fmap2, fred2)
success = job.waitForCompletion()
```

### 3) Todos los jugadores que “retaron” a más de **H** oponentes **distintos**

*(H es un parámetro de la consulta).*

---

**Explicación**

Este job identifica a los jugadores que han retado a más de **H** oponentes diferentes.

- En la **fase Map**, se emiten pares con el ID del retador y del oponente retado.  
- En la **fase Reduce**, se agrupan todos los oponentes retados por cada jugador y se cuentan los **distintos**.  
- Si la cantidad supera el umbral `H`, el jugador se considera parte del resultado.

De esta forma se obtiene la lista de jugadores con mayor diversidad de oponentes en sus combates.

### DAG



```python
import sys
sys.path.append("..") 
from MRE import Job

inputDir = "../Datasets/TP1/input/"
outputDir = "../Datasets/TP1/out_punto3/"

# Parámetro H
H = 12

def fmap3(key, value, context):
    cols = value.strip().split()
    id_retador = key
    id_retado  = cols[0]
    context.write(id_retador, id_retado)

def fred3(key, values, context):
    oponentes = set()
    for v in values:
        oponentes.add(v)
    cantidad = len(oponentes)
    context.write("DISTINTOS", (key, cantidad))

job = Job(inputDir, outputDir, fmap3, fred3)
success = job.waitForCompletion()
```

### Procedimiento que no es big data

```python
import os
jugadores = []
with open(os.path.join(outputDir, "output.txt")) as f:
    for line in f:
        _, jugador, cantidad = line.strip().split("\t")
        cantidad = int(cantidad)
        if cantidad >= H:
            jugadores.append(jugador)

print("\nConsulta 3:")
print(f"Con H = {H}")
print("Jugadores que retaron a más de H oponentes distintos:", jugadores)
```

### 4) El top 10 de los jugadores con mejor puntaje heroico

El puntaje **PH** de un jugador *i* se calcula como:

$$
PH_i = \alpha \left( \sum_{j=1}^{n} PH_j \cdot \frac{PP_i}{PP_j} \right) + (1 - \alpha)
$$

donde:

- *n* es la cantidad de jugadores “retados” por el jugador *i*.  
- *PH_j* es el puntaje **P** del *j*-ésimo jugador retado por *i*.  
- *PP_i* es el puntaje promedio del jugador *i*.  
- *PP_j* es el puntaje promedio del *j*-ésimo jugador retado por *i*.  
- *α* es un parámetro del algoritmo.

<details><summary>Nota sobre el cálculo del puntaje heroico (PH)</summary>

Como para calcular el puntaje **PH** de un jugador *i* previamente hay que calcular el puntaje de los jugadores a quiénes retó (y a su vez éstos pueden haber retado al jugador *i*), el cálculo se realiza de manera **iterativa**.

Se comienza asignando el mismo puntaje heroico con un valor arbitrario a todos los jugadores (**tiempo 0**).  
Con esos puntajes arbitrarios se calcula el puntaje usando la fórmula detallada para obtener los nuevos puntajes (**tiempo 1**),  
con éstos se calculan los nuevos puntajes (**tiempo 2**) y así se continúa hasta alcanzar una **convergencia en los valores**.

# Implementación MapReduce: Cálculo del Puntaje Heroico

En este trabajo se aplica el paradigma **MapReduce** para analizar información de combates entre jugadores de un videojuego online.  
Se implementan tres *jobs* secuenciales:

1. **Job 1 – Filtrado de datos**  
   Limpia las líneas de entrada y deja solo aquellas con formato válido.

2. **Job 2 – Cálculo del Promedio de Puntaje (PP)**  
   Calcula el puntaje promedio obtenido por cada jugador retador.

3. **Job 3 – Cálculo del Puntaje Heroico (PH)**  
   Itera aplicando una fórmula de propagación de prestigio entre jugadores hasta alcanzar convergencia.

Finalmente, se muestra el **Top 10 de jugadores más heroicos**.


```python
import sys, os, re
sys.path.append("..")
from MRE import Job

BASE_DIR = os.path.join(os.getcwd(), "Datasets", "TP1")

input_path  = os.path.join(BASE_DIR, "input")
job1_out    = os.path.join(BASE_DIR, "out_filtrar")
job2_out    = os.path.join(BASE_DIR, "out_pp")
job3_out    = os.path.join(BASE_DIR, "out_ph")

# Crear las carpetas de salida si no existen
for path in [input_path, job1_out, job2_out, job3_out]:
    os.makedirs(path, exist_ok=True)

print("Estructura de carpetas lista")
```

## Job 1 – Filtrado de datos

El objetivo de este primer *job* es leer los combates desde el archivo `jugadores.txt`,  
verificar que cada línea tenga los tres campos esperados (`retador`, `puntos`, `tiempo`)  
y emitir solo los registros válidos para procesar en las etapas siguientes.


```python
def fmap_filtrar(key, value, context):
    parts = value.strip().split()
    retador, puntos, tiempo = parts
    puntos, tiempo = float(puntos), float(tiempo)
    context.write(retador, (puntos, tiempo))

def fred_filtrar(key, values, context):
    for v in values:
        context.write(key, v)

print("Funciones Map y Reduce del Job 1 definidas correctamente")
```

## Job 2 – Cálculo del promedio de puntaje (PP)

Cada jugador puede participar en múltiples combates.  
En este *job* se calcula su puntaje promedio acumulando las puntuaciones individuales.

```python
def fmap_pp(key, value, context):
    if isinstance(value, tuple):
        puntos, tiempo = value
    else:
        puntos, tiempo = map(float, value.strip().split())
    context.write(key, ("P", puntos))

def fred_pp(key, values, context):
    total_p, n = 0.0, 0
    for tag, val in values:
        if tag == "P":
            total_p += val
            n += 1
    if n > 0:
        promedio = total_p / n
        context.write(key, promedio)

print("Funciones Map y Reduce del Job 2 definidas correctamente")
```

## Job 3 – Cálculo del Puntaje Heroico (PH)

El puntaje heroico se calcula iterativamente aplicando una fórmula de propagación basada en el *prestigio* de los oponentes.

La idea es que un jugador tiene mayor **PH** si derrota a oponentes con alto PH y con mejor PP (puntaje promedio).

Se utiliza la siguiente fórmula:

$$
PH_{nuevo}(i) = \alpha \sum_j \left( PH_{previo}(j) \times \frac{PP(i)}{PP(j)} \right) + (1 - \alpha)
$$


```python
def fmap_ph(retador, value, context):
    parts = re.split(r'[\t\s]+', value.strip())
    retado, puntos, tiempo = map(float, parts)
    retador = int(retador)
    retado = int(retado)

    pp = context["pp"]
    ph_prev = context["ph_prev"]
    alpha = context["alpha"]

    if retador not in pp or retado not in pp or retado not in ph_prev:
        return
    if pp[retado] == 0:
        return

    contrib = ph_prev[retado] * (pp[retador] / pp[retado])
    context.write(retador, contrib)

def fred_ph(jugador, values, context):
    vals = list(values)
    if not vals:
        print(f"[REDUCE][JOB3] Jugador {jugador} sin contribuciones.")
        return

    total = sum(vals)
    alpha = context["alpha"]
    nuevo_ph = alpha * total + (1 - alpha)

    context.write(jugador, nuevo_ph)

print("Funciones Map y Reduce del Job 3 definidas correctamente")
```

## Funciones auxiliares

Estas funciones permiten leer los resultados de cada job,  
procesar los archivos de salida y generar el Top 10 final.

```python
def leer_resultados(path):
    res = {}
    try:
        archivos = [f for f in os.listdir(path) if f.startswith("part") or f.startswith("output")]
        if not archivos:
            print(f"[WARN] No se encontró ningún archivo en {path}")
            return res

        file_path = os.path.join(path, archivos[0])
        with open(file_path, "r", encoding="utf-8") as f:
            for line in f:
                line = line.strip()
                if not line or "\t" not in line:
                    continue
                k, v = line.split("\t", 1)
                try:
                    res[int(k)] = float(v.strip().replace("(", "").replace(")", "").replace(",", ""))
                except ValueError:
                    continue
    except FileNotFoundError:
        print(f"[WARN] Carpeta no encontrada: {path}")
    print(f"[DEBUG] Leídos {len(res)} resultados desde {path}")
    return res

def guardar_top10(top10, path):
    out_path = os.path.join(path, "top10.txt")
    with open(out_path, "w", encoding="utf-8") as f:
        for j, p in top10:
            f.write(f"{j}\t{p:.2f}\n")
```

## 🚀 Ejecución del pipeline completo

```python
print(f"\n[DEBUG] Archivos en {input_path}: {os.listdir(input_path)}")

# JOB 1
print("\n========== INICIANDO JOB 1: FILTRADO ==========")
job1 = Job(input_path, job1_out, fmap_filtrar, fred_filtrar)
job1.waitForCompletion()
print("[OK] Job 1 completado")

# JOB 2
print("\n========== INICIANDO JOB 2: PROMEDIO DE PUNTAJES ==========")
job2 = Job(job1_out, job2_out, fmap_pp, fred_pp)
job2.waitForCompletion()
pp = leer_resultados(job2_out)
print(f"[OK] Job 2 completado | Jugadores con PP calculado: {len(pp)}")

# JOB 3
print("\n========== INICIANDO JOB 3: PUNTAJE HEROICO ==========")
jugadores = set(pp.keys())
with open(os.path.join(input_path, "jugadores.txt"), "r", encoding="utf-8") as f:
    for linea in f:
        partes = linea.strip().split()
        if len(partes) == 4:
            jugadores.add(int(partes[1]))
ph = {j: 1.0 for j in jugadores}

alpha, error, max_iter = 0.202, 0.1, 20

for i in range(max_iter):
    print(f"\n---- Iteración {i+1} ----")
    job3 = Job(input_path, job3_out, fmap_ph, fred_ph)
    job3.setParams({"pp": pp, "ph_prev": ph, "alpha": alpha})
    job3.waitForCompletion()

    nuevo_ph = leer_resultados(job3_out)
    if not nuevo_ph:
        print("[WARN] No se obtuvieron nuevos PH")
        break

    dif = sum(abs(nuevo_ph[j] - ph.get(j, 0)) for j in nuevo_ph) / len(nuevo_ph)
    print(f"[INFO] Iter {i+1} | Diferencia promedio Δ={dif:.4f}")

    ph = nuevo_ph
    if dif < error:
        print(f"[INFO] ✅ Convergencia alcanzada en {i+1} iteraciones.")
        break

print("\n========== TOP 10 FINAL ==========")
top10 = sorted(ph.items(), key=lambda x: x[1], reverse=True)[:10]
for j, p in top10:
    print(f"(Jugador = {j}, PH = {p:.2f})")
guardar_top10(top10, job3_out)
print(f"[OK] Resultados guardados en: {job3_out}")
```
